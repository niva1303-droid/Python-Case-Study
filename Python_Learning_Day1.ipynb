# Case Study – Library Fine Calculator

# Instructions:

1. Create variables:

student_name

book_name

days_late

fine_per_day

2. Use input() function to get details from the user.

3. Use type casting:

int() for days_late

float() for fine_per_day

4. Calculate:

total_fine = days_late * fine_per_day

5. Display all details using:

print()

f-string

6. Display datatype of important variables using:

type()

```python
# Library Fine Calculator

student_name = input("Enter Student Name : ")
book_name = input("Enter Book Name : ")
days_late = int(input("Number of Days Late : "))
fine_per_day = float(input("Fine Per Day : "))

# Calculate

total_fine = days_late * fine_per_day

# Display details

print("\n---Library Fine Calculator---")
print(f"Student Name: {student_name}")
print(f"Book Name: {book_name}")
print(f"Number of Days Late: {days_late}")
print(f"Fine Per Day: Rs. {fine_per_day}")
print(f"Total Fine: Rs. {total_fine:.2f}")

# Data Types

print("\n---Data Types---")
print(f"Student Name: {type(student_name)}")
print(f"Book Name: {type(book_name)}")
print(f"Number of Days Late: {type(days_late)}")
print(f"Fine Per Day: {type(fine_per_day)}")
print(f"Total Fine: {type(total_fine)}")

```
