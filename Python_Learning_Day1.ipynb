# Case Study – Student Registration System

**Instructions:**

1. Create variables:

student_name

student_age

course_name

course_fee

2. Use input() function to get student details from the user.

3. Use type casting:

int() for student age

float() for course fee

4. Calculate:

gst = 18% of course fee

total_fee = course fee + gst

5. Display all student details using:

print()

f-string

6. Display datatype of important variables using:

type()

```python
# Student Registration System

student_name = input("Enter Student Name: " )
student_age = int(input("Enter Student Age: "))
course_name = input("Enter Course Name: ")
course_fee = float(input("Enter Course Fee: "))

# GST and Total Fee

GST = 0.18*(course_fee)
total_fee = course_fee + GST

# Display details

print("\n---Student Registration Details---")
print(f"Student Name: {student_name}")
print(f"Student Age: {student_age}")
print(f"Course Name: {course_name}")
print(f"Course Fee: Rs. {course_fee:.2f}")
print(f"GST: Rs.{GST:.2f}")
print(f"Total Fee: Rs. {total_fee:.2f}")

# Data Types
print("\n---Data Types---")
print(f"Student Name: {type(student_name)}")
print(f"Student Age: {type(student_age)}")
print(f"Course Name: {type(course_name)}")
print(f"Course Fee: {type(course_fee)}")
print(f"GST: {type(GST)}")
print(f"Total Fee: {type(total_fee)}")
```
